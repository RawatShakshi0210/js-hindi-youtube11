```python
def simple_chatbot():
  print("Hello! I'm a simple chatbot. Type 'quit' to exit.")
  while True:
    user_input = input("You: ").lower()
    if user_input.lower() == 'quit':
      print("Chatbot: Goodbye!")
      break
    elif 'hello' in user_input or 'hi' in user_input:

      print("Chatbot: Hi there!How can i help you today?")
    elif 'how are you' in user_input:
      print("Chatbot: I'm just a chatbot, but I'm here to help!")
    elif 'name' in user_input:
      print("Chatbot: I don't have a name, but you can call me Chatbot!")
    elif 'weather' in user_input:
      print("Chatbot: I cannot tell you about the weather, as i don't have access to real-time data.")
    elif 'help' in user_input:
      print("Chatbot:I can answer basic questions or just chat.Try asking me about my name!")
    elif 'created' in  user_input:
      print("Chatbot:Who is my creater.Shakshi Rawat!")
    else:
      print("Chatbot:I'm sorry,I don,t understand that.Can you rephrase?")
```

```python
simple_chatbot()
```

```python

conservation_history = [
    "Chatbot: Hello! I'm a simple chatbot. Type 'quit' to exit.",
    "You: Hello",
    "Chatbot: Hi there!How can i help you today?",
    "You: How are you?",
    "Chatbot: I'm just a chatbot, but I'm here to help!",
    "You: What's your name?",
    "Chatbot: I don't have a name, but you can call me Chatbot!",
    "You:quit"
    "Chatbot:Goodbye!"

]
```

```python
# SAVE CONSERVATION TO TEXT FILE
filename = "chatbot_conservation.txt"

try:
  with open(filename,"w") as f:
    for line in conservation_history:
      f.write(line + "\n")
  print(f"Conservation history saved to {filename}")
except IO Error as e:
  print(f"Error saving conservation to file: {e}")
```

```python
try:
  with open(filename,"r") as f:
    content = f.read()
    print("\n--- Contents of" + filename + " ---\n")
    print(content)
 print("\n----------------------------")
except FileNotFoundError:
  print(f"Error:The file{filename}was not found")
except IOError as e:
  print(f"Error reading file: {e}")
```

```python
# NEW CHAT ENTRIES TO APPEND

new_chat_entries = [
    "You:Tell me a joke.",
    "Chatbot:Why don't scientists trust atoms?Because they make up everything!",
    "You:That's a good one!"
]
```

```python
# NEW CHAT ENTRIES TO APPEND

try:
  with open(filename,"a") as f:
    for entry in new_chat_entries:
      f.write(entry + "\n")
  print(f"Successfully appended new entries to {filename}")
except IOError as e:
  print(f"Error appending to file:{e}")
```

```python
# VERIFY APPENDED CONTENT

try:
  with open(filename,"r") as f:
    updated_content = f.read()
    print("\n--- Updated Contents of" + filename + " ---\n")
    print(updated_content)
 print("\n----------------------------")
except FileNotFoundError:
  print(f"Error:The file{filename}was not found")
except IOError as e:
  print(f"Error reading file: {e}")
```
